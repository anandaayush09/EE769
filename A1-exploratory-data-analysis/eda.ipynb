```python
#[2]
#[3]

# i have taken a sample of 4500 rows because we cannot operate with more than 5000 rows in altair

df2_sampled = df2.sample(4500)


df2_sampled['year'] = df2_sampled['activity_date'].dt.year

# Create a histogram of the `score` column
score_histogram = alt.Chart(df2_sampled).mark_bar().encode(
    x=alt.X('score:Q', bin=True, title='Score'),
    y=alt.Y('count()', title='Number of Inspections'),
    tooltip=[alt.Tooltip('score:Q', bin=True, title='Score'), 'count()']
).properties(
    title='Inspection Scores'
).interactive()

# Create a bar chart showing the count of each `grade`
grade_bar_chart = alt.Chart(df2_sampled).mark_bar().encode(
    x=alt.X('grade:N', sort='-y', title='Grade'),
    y=alt.Y('count()', title='Number of Inspections'),
    color='grade',
    tooltip=['grade', 'count()']
).properties(
    title='Inspections by Grade'
)

# Group the data by `year` and calculate the mean `score` for each year
yearly_avg_score = df2_sampled.groupby('year')['score'].mean().reset_index()

# Create a line chart to visualize the trend of average `score` over time
avg_score_trend = alt.Chart(yearly_avg_score).mark_line(point=True).encode(
    x='year:O',
    y='score:Q',
    tooltip=['year', 'score']
).properties(
    title='Average Inspection Score Trend Over Time'
).interactive()

# Display the charts
score_histogram.save('inspection_score_histogram.json')
grade_bar_chart.save('inspection_count_by_grade_bar_chart.json')
avg_score_trend.save('average_inspection_score_trend_line_chart.json')
```

```python
# Extracting ZIP codes from the column "SITE ADDRESS" of df1 and "facility_zip" of df2

df1['SITE ADDRESS'] = df1['SITE ADDRESS'].str.extract(r'(\d{5})$')
df2['facility_zip'] = df2['facility_zip'].str.extract(r'^(\d{5})')
```

```python
#[4]

engine = sqlalchemy.create_engine('sqlite:///:memory:')

df1.to_sql('df1', engine, index=False)
df2.to_sql('df2', engine, index=False)

query_df1 = """
SELECT
    `SITE ADDRESS`,
    COUNT(*) AS total_records,
    AVG(`JOBS`) AS average_jobs,
    MAX(`JOBS`) AS max_jobs,
    MIN(`JOBS`) AS min_jobs,
    AVG(`SITE UNITS`) AS average_site_units,
    MAX(`SITE UNITS`) AS max_site_units,
    MIN(`SITE UNITS`) AS min_site_units,
    MAX(`HOUSING TYPE`) AS max_housing_type,
    MIN(`HOUSING TYPE`) AS min_housing_type
FROM
    df1
GROUP BY
    `SITE ADDRESS`
"""

result_df1 = pd.read_sql_query(query_df1, engine)

query_df2 = """
SELECT
    facility_zip,
    COUNT(*) AS total_records,
    AVG(score) AS average_score,
    MAX(score) AS max_score,
    MIN(score) AS min_score,
    MAX(service_description) AS max_service_description,
    MIN(service_description) AS min_service_description
FROM
    df2
GROUP BY
    facility_zip
"""

result_df2 = pd.read_sql_query(query_df2, engine)

print("Summary of df1")
print(result_df1)

print("\nSummary of df2")
print(result_df2)
```

```python
# We will fill all these missing values with the help of linear interpolation

df3.interpolate(method='linear', inplace=True)
missing_values_df3 = df3.isnull().sum()
print(missing_values_df3.to_markdown(numalign="left", stralign="left"))
```

```python
# Now we plot some graphs to get some useful insights from the data

print(f"Unique values in Month column: {df3['Month'].unique()}")
```

```python
#[2]
#[3]

# there are some abnormal values like "Marcrh" and "November with a trailing space so we need to clear these values"

# Replace incorrect month names
df3['Month'] = df3['Month'].replace({'November ': 'November', 'Marcrh': 'March'})

# Combine `Year` and `Month` to create a `Date` column and convert it to datetime
df3['Date'] = pd.to_datetime(df3['Year'].astype(str) + ' ' + df3['Month'], format='%Y %B')

# Create a line chart to visualize the trend of the `General index` over time
line_chart = alt.Chart(df3).mark_line(point=True).encode(
    x='Date:T',
    y='General index:Q',
    tooltip = ['Date', 'General index']
).properties(
    title='General Index Trend Over Time'
).interactive()

# Create a bar chart to compare the average `General index` across different `Sectors`
bar_chart = alt.Chart(df3).mark_bar().encode(
    x='Sector:N',
    y='mean(General index):Q',
    color='Sector:N',
    tooltip = ['Sector', alt.Tooltip('mean(General index)', title='Avg. General Index')]
).properties(
    title='Average General Index by Sector'
)

# Create a grouped bar chart to show the average `General index` for each `Sector` over the years
grouped_bar_chart = alt.Chart(df3).mark_bar().encode(
    x='Year:O',
    y='mean(General index):Q',
    color='Sector:N',
    column='Sector:N',
    tooltip = ['Year', 'Sector', alt.Tooltip('mean(General index)', title='Avg. General Index')]
).properties(
    title='Average General Index by Sector and Year'
)

# Display the charts
line_chart.save('general_index_trend_over_time_line_chart.json')
bar_chart.save('average_general_index_by_sector_bar_chart.json')
grouped_bar_chart.save('average_general_index_by_sector_and_year_grouped_bar_chart.json')
```
